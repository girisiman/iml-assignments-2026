## <font face="Georgia" size="5" color="black"> Task 1 - Compute Ordinary Least Squares Solution

<font face="Georgia" size="3" color="black">

Linear Regression models the relationship between an independent variable $\mathbf{x}$ and a dependent variable $\mathbf{y}$ using a straight line.

The regression equation is

$$
\boxed{
y=\beta_0+\beta_1x
}
$$

where

- $\mathbf{\beta_0}$ is the intercept,
- $\mathbf{\beta_1}$ is the slope,
- $\mathbf{x}$ is the independent variable,
- $\mathbf{y}$ is the predicted value.

---

</font>

<font face="Georgia" size="3" color="black">


## Step 1: Define the Cost Function

The Ordinary Least Squares (OLS) method estimates the parameters by minimizing the sum of squared errors (SSE),

$$
\boxed{
J(\beta_0,\beta_1)
=
\sum_{i=1}^{n}
\left(y_i-(\beta_0+\beta_1x_i)\right)^2.
}
$$

---

## Step 2: Differentiate with Respect to the Parameters

Taking the partial derivative with respect to the intercept,

$$
\frac{\partial J}{\partial \beta_0}
=
-2
\sum_{i=1}^{n}
\left(
y_i-\beta_0-\beta_1x_i
\right)
=0.
$$

Taking the partial derivative with respect to the slope,

$$
\frac{\partial J}{\partial \beta_1}
=
-2
\sum_{i=1}^{n}
x_i
\left(
y_i-\beta_0-\beta_1x_i
\right)
=0.
$$

These equations are known as the **Normal Equations**.

---

## Step 3: Solve the Normal Equations

Solving the above equations gives the Ordinary Least Squares estimates

$$
\boxed{
\beta_1=
\frac{
n\sum x_i y_i-\sum x_i\sum y_i
}{
n\sum x_i^2-(\sum x_i)^2
}
}
$$

and

$$
\boxed{
\beta_0=
\bar y-\beta_1\bar x
}
$$

where

- $\bar x$ is the mean of the input values,
- $\bar y$ is the mean of the output values.

---

## Final Result

The Ordinary Least Squares solution estimates the slope and intercept of the regression line by minimizing the sum of squared prediction errors.

The obtained formulas are

$$
\boxed{
\beta_1=
\frac{
n\sum x_i y_i-\sum x_i\sum y_i
}{
n\sum x_i^2-(\sum x_i)^2
}
}
$$

and

$$
\boxed{
\beta_0=
\bar y-\beta_1\bar x.
}
$$

These equations will be used in the next task to compute the slope and intercept for the given house price dataset.

## Implementation with Normal Equation

</font>

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# ---------------------------------------------
# House Price Dataset
# ---------------------------------------------
X = np.array([600, 800, 1000, 1200, 1400, 1600]).reshape(-1, 1)
y = np.array([45, 58, 72, 85, 96, 110])

# =============================================
# Method 1: Normal Equation
# =============================================

# Add a column of ones for the intercept
X_b = np.c_[np.ones((X.shape[0], 1)), X]

# Compute parameters using the Normal Equation
theta = np.linalg.inv(X_b.T @ X_b) @ X_b.T @ y

intercept_ne = theta[0]
slope_ne = theta[1]

print("="*60)
print("Normal Equation")
print("="*60)
print(f"Intercept (β₀): {intercept_ne:.6f}")
print(f"Slope (β₁):     {slope_ne:.6f}")
```

```python
# Predictions
y_pred_ne = X_b @ theta

# Prediction for 1500 sq. ft.
price_1500_ne = intercept_ne + slope_ne * 1500

print(f"\nPredicted price for 1500 sq. ft.: {price_1500_ne:.3f} NPR Lakhs")
```

```python
# Performance metrics
mse_ne = mean_squared_error(y, y_pred_ne)
r2_ne = r2_score(y, y_pred_ne)

print(f"MSE : {mse_ne:.6f}")
print(f"R²  : {r2_ne:.6f}")
```

<font face="Georgia" size="3" color="black">

## Implementation with scikitlearn

</font>

```python
# =============================================
# Method 2: Scikit-learn Linear Regression
# =============================================

model = LinearRegression()
model.fit(X, y)

intercept_sk = model.intercept_
slope_sk = model.coef_[0]

print("\n" + "="*60)
print("Scikit-learn Linear Regression")
print("="*60)

print(f"Intercept (β₀): {intercept_sk:.6f}")
print(f"Slope (β₁):     {slope_sk:.6f}")
```

```python
# Predictions
y_pred_sk = model.predict(X)

# Prediction for 1500 sq. ft.
price_1500_sk = model.predict([[1500]])[0]

print(f"\nPredicted price for 1500 sq. ft.: {price_1500_sk:.3f} NPR Lakhs")
```

```python
# Performance metrics
mse_sk = mean_squared_error(y, y_pred_sk)
r2_sk = r2_score(y, y_pred_sk)

print(f"MSE : {mse_sk:.6f}")
print(f"R²  : {r2_sk:.6f}")
```
