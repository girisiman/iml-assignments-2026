The class mean is the average of all samples belonging to a class. It is computed using

$$\mathbf{
\boxed{

=
\frac{1}{n}
\sum_{i=1}^{n}\mathbf{x}_i
}

where

- $n$ is the number of samples in the class,
- $\mathbf{x}_i$ is the $i^{\text{th}}$ sample.

## <font face="Georgia" size="5" color="black"> Task 2 - Compute Compute within class scatter matrix

<font face="Georgia" size="3" color="black">
The within-class scatter matrix measures the spread of samples around their respective class means. It is computed as the sum of the scatter matrices of all classes.

The within-class scatter matrix is given by

$$\mathbf{
\boxed{
\mathbf{S}_W=\mathbf{S}_1+\mathbf{S}_2
}
}$$

where

- $\mathbf{S}_1$ is the scatter matrix of the Healthy class,
- $\mathbf{S}_2$ is the scatter matrix of the Diseased class.

For each class,

$$\mathbf{
\boxed{
\mathbf{S}
=
\sum_{i=1}^{n}
(\mathbf{x}_i-\boldsymbol{\mu})
(\mathbf{x}_i-\boldsymbol{\mu})^T
}
}$$

where

- $\mathbf{x}_i$ is a sample,
- $\boldsymbol{\mu}$ is the class mean.

---

</font>


<font face="Georgia" size="3" color="black">


## Step 1: Compute the Scatter Matrix of the Healthy Class

From Task 1, the mean vector of the Healthy class is

$$
\boldsymbol{\mu}_1=
\begin{bmatrix}
3\\
3
\end{bmatrix}.
$$

Arrange the Healthy class samples into a data matrix,

$$
\mathbf{X}_1=
\begin{bmatrix}
2 & 3\\
3 & 4\\
4 & 3\\
3 & 2
\end{bmatrix}.
$$

Since the mean vector must be subtracted from every observation, we will follow <b>the vectorization trick</b>  i.e. repeat the mean vector for all four samples,

$$
\mathbf{M}_1=
\begin{bmatrix}
3 & 3\\
3 & 3\\
3 & 3\\
3 & 3
\end{bmatrix}.
$$

Subtract the mean matrix from the data matrix,

$$
\begin{aligned}
\mathbf{A}
&=\mathbf{X}_1-\mathbf{M}_1\\
&=
\begin{bmatrix}
2 & 3\\
3 & 4\\
4 & 3\\
3 & 2
\end{bmatrix}
-
\begin{bmatrix}
3 & 3\\
3 & 3\\
3 & 3\\
3 & 3
\end{bmatrix}\\
&=
\begin{bmatrix}
2-3 & 3-3\\
3-3 & 4-3\\
4-3 & 3-3\\
3-3 & 2-3
\end{bmatrix}\\
&=
\begin{bmatrix}
-1 & 0\\
0 & 1\\
1 & 0\\
0 & -1
\end{bmatrix}.
\end{aligned}
$$

The scatter matrix of the Healthy class is

$$
\mathbf{S}_1=\mathbf{A}^{T}\mathbf{A}.
$$

Substituting,

$$
\begin{aligned}
\mathbf{S}_1
&=
\begin{bmatrix}
-1&0&1&0\\
0&1&0&-1
\end{bmatrix}
\begin{bmatrix}
-1&0\\
0&1\\
1&0\\
0&-1
\end{bmatrix}\\
&=
\boxed{
\begin{bmatrix}
2&0\\
0&2
\end{bmatrix}.
}
\end{aligned}
$$

## <font face="Georgia" size="5" color="black"> Task 3 - Compute between-class scatter matrix

<font face="Georgia" size="3" color="black">

The between-class scatter matrix measures the separation between the class means. It is defined as

$$
\boxed{
\mathbf{S}_B
=
\sum_{i=1}^{c}
n_i
(\boldsymbol{\mu}_i-\boldsymbol{\mu})
(\boldsymbol{\mu}_i-\boldsymbol{\mu})^T
}
$$

where

- $c$ is the number of classes,
- $n_i$ is the number of samples in class $i$,
- $\boldsymbol{\mu}_i$ is the mean of class $i$,
- $\boldsymbol{\mu}$ is the overall mean of the dataset.

---

</font>

<font face="Georgia" size="3" color="black">

## Step 1: Compute the Overall Mean

From Task 1,

$$
\boldsymbol{\mu}_1=
\begin{bmatrix}
3\\
3
\end{bmatrix},
\qquad
\boldsymbol{\mu}_2=
\begin{bmatrix}
8\\
8
\end{bmatrix}.
$$

Since each class contains four samples, the overall mean is

$$
\begin{aligned}
\boldsymbol{\mu}
&=
\frac{4\boldsymbol{\mu}_1+4\boldsymbol{\mu}_2}{8}\\
&=
\frac{1}{2}
\left(
\begin{bmatrix}
3\\
3
\end{bmatrix}
+
\begin{bmatrix}
8\\
8
\end{bmatrix}
\right)\\
&=
\boxed{
\begin{bmatrix}
5.5\\
5.5
\end{bmatrix}.
}
\end{aligned}
$$

---
## Step 2: Compute the Scatter for the Healthy Class

Compute the deviation of the Healthy class mean from the overall mean,

$$
\begin{aligned}
\boldsymbol{\mu}_1-\boldsymbol{\mu}
&=
\begin{bmatrix}
3\\
3
\end{bmatrix}
-
\begin{bmatrix}
5.5\\
5.5
\end{bmatrix}\\
&=
\begin{bmatrix}
-2.5\\
-2.5
\end{bmatrix}.
\end{aligned}
$$

Compute the outer product,

$$
\begin{aligned}
(\boldsymbol{\mu}_1-\boldsymbol{\mu})
(\boldsymbol{\mu}_1-\boldsymbol{\mu})^T
&=
\begin{bmatrix}
-2.5\\
-2.5
\end{bmatrix}
\begin{bmatrix}
-2.5 & -2.5
\end{bmatrix}\\
&=
\begin{bmatrix}
6.25 & 6.25\\
6.25 & 6.25
\end{bmatrix}.
\end{aligned}
$$

Since there are four Healthy samples,

$$
\boxed{
4
\begin{bmatrix}
6.25 & 6.25\\
6.25 & 6.25
\end{bmatrix}
=
\begin{bmatrix}
25 & 25\\
25 & 25
\end{bmatrix}.
}
$$

---

## Step 3: Compute the Scatter for the Diseased Class

\begin{bmatrix}

Since there are four Diseased samples,

## <font face="Georgia" size="5" color="black"> Task 4 - Determine optimal LDA projection vector

<font face="Georgia" size="3" color="black">

The objective of Linear Discriminant Analysis (LDA) is to determine a projection vector that maximizes the separation between classes while minimizing the spread of samples within each class.

The optimal projection vector is given by

$$
\boxed{
\mathbf{w}
=
\mathbf{S}_W^{-1}
(\boldsymbol{\mu}_2-\boldsymbol{\mu}_1)
}
$$

where

- $\mathbf{S}_W$ is the within-class scatter matrix,
- $\boldsymbol{\mu}_1$ and $\boldsymbol{\mu}_2$ are the class mean vectors.

---

</font>

<font face="Georgia" size="3" color="black">

## Step 1: Compute the Difference Between the Class Means

From Task 1,

$$
\boldsymbol{\mu}_1=
\begin{bmatrix}
3\\
3
\end{bmatrix},
\qquad
\boldsymbol{\mu}_2=
\begin{bmatrix}
8\\
8
\end{bmatrix}.
$$

Therefore,

$$
\begin{aligned}
\boldsymbol{\mu}_2-\boldsymbol{\mu}_1
&=
\begin{bmatrix}
8\\
8
\end{bmatrix}
-
\begin{bmatrix}
3\\
3
\end{bmatrix} \\
&=
\boxed{
\begin{bmatrix}
5\\
5
\end{bmatrix}.
}
\end{aligned}
$$

---

## Step 2: Compute the Inverse of the Within-Class Scatter Matrix

From Task 2,

$$
\mathbf{S}_W=
\begin{bmatrix}
4&0\\
0&4
\end{bmatrix}.
$$

Since this is a diagonal matrix, its inverse is obtained by taking the reciprocal of each diagonal element,

$$
\boxed{
\mathbf{S}_W^{-1}
=
\begin{bmatrix}
\frac14&0\\
0&\frac14
\end{bmatrix}.
}
$$

---

## Step 3: Compute the Projection Vector

Substitute the values into the LDA equation,

$$
\begin{aligned}
\mathbf{w}
&=
\mathbf{S}_W^{-1}
(\boldsymbol{\mu}_2-\boldsymbol{\mu}_1)\\
&=
\begin{bmatrix}
\frac14&0\\
0&\frac14
\end{bmatrix}
\begin{bmatrix}
5\\
5
\end{bmatrix}.
\end{aligned}
$$

Perform the matrix multiplication,

$$
\begin{aligned}
\mathbf{w}
&=
\begin{bmatrix}
\frac14(5)+0(5)\\
0(5)+\frac14(5)
\end{bmatrix}\\
&=
\boxed{
\begin{bmatrix}
1.25\\
1.25
\end{bmatrix}.
}
\end{aligned}
$$

---

## Step 4: Normalize the Projection Vector

The direction of the projection vector is more important than its magnitude. Therefore, it is common to normalize the vector to unit length.

The magnitude of the vector is

$$
\begin{aligned}
\|\mathbf{w}\|
&=
\sqrt{1.25^2+1.25^2}\\
&=
\sqrt{3.125}\\
&\approx1.768.
\end{aligned}
$$

Hence, the normalized projection vector is

$$
\begin{aligned}
\hat{\mathbf{w}}
&=
\frac{\mathbf{w}}{\|\mathbf{w}\|}\\
&=
\frac{1}{1.768}
\begin{bmatrix}
1.25\\
1.25
\end{bmatrix}\\
&\approx
\boxed{
\begin{bmatrix}
0.707\\
0.707
\end{bmatrix}.
}
\end{aligned}
$$

---

## Final Answer

The optimal LDA projection vector is

$$
\boxed{
\mathbf{w}
=
\begin{bmatrix}
1.25\\
1.25
\end{bmatrix}.
}
$$

The corresponding unit projection vector is

$$
\boxed{
\hat{\mathbf{w}}
=
\begin{bmatrix}
0.707\\
0.707
\end{bmatrix}.
}
$$

The projection vector points in the direction that best separates the Healthy and Diseased classes while minimizing the spread within each class. Since the vector has equal components in both dimensions, the LDA projection axis lies along the line \(y=x\).

## <font face="Georgia" size="5" color="black"> Task 6 - Discuss linear separability

<font face="Georgia" size="3" color="black">

Linear separability refers to the ability to separate different classes using a single linear decision boundary (or threshold).

From Task 5, the projected values on the LDA axis are

- **Healthy Class:** $\mathbf{3.535,\ 4.949,\ 4.949,\ 3.535}$
- **Diseased Class:** $\mathbf{10.605,\ 12.019,\ 12.019,\ 10.605}$

The largest projected value for the Healthy class is

$$
\boxed{4.949}
$$

while the smallest projected value for the Diseased class is

$$
\boxed{10.605.}
$$

Since

$$
4.949 < 10.605,
$$

there is a clear gap between the projected values of the two classes. Therefore, a single threshold can be chosen between these values to correctly classify every sample.

For example, choosing a threshold of

$$
\boxed{t = 7.777}
$$

(which is the midpoint between ($4.949$ and $10.605$) gives the decision rule

$$
\begin{cases}
y < 7.777, & \text{Healthy Class},\\
y \ge 7.777, & \text{Diseased Class}.
\end{cases}
$$

Using this threshold, all Healthy samples lie on one side of the decision boundary, while all Diseased samples lie on the other side.

## Conclusion

The two classes are **perfectly linearly separable** after projection onto the LDA axis. No projected samples overlap, indicating that the LDA projection successfully maximizes class separation for this dataset

## <font face="Georgia" size="5" color="black"> Task 7 - Compare PCA and LDA

<font face="Georgia" size="3" color="black">

Although both Principal Component Analysis (PCA) and Linear Discriminant Analysis (LDA) are dimensionality reduction techniques, they have different objectives and use different criteria for selecting projection directions.

| Principal Component Analysis (PCA) | Linear Discriminant Analysis (LDA) |
|------------------------------------|------------------------------------|
| Unsupervised learning technique. | Supervised learning technique. |
| Does not require class labels. | Requires class labels. |
| Maximizes the variance of the projected data. | Maximizes the separation between different classes while minimizing the spread within each class. |
| Projection directions are called **principal components**. | Projection direction is called the **linear discriminant**. |
| Components are obtained from the eigenvectors of the covariance matrix. | Projection vector is computed using the within-class and between-class scatter matrices. |
| Useful for data compression, visualization, and noise reduction. | Useful for feature extraction and classification. |
| Can produce up to the number of original features (or one less after centering). | For \(c\) classes, at most \(c-1\) discriminant directions can be obtained. |

## Comparison for This Assignment

In **Question 1 (PCA)**, the projection directions were chosen to preserve the maximum variance in the employee performance dataset. The first principal component explained almost all of the total variance, allowing the three-dimensional data to be reduced to two dimensions with minimal information loss.

In **Question 2 (LDA)**, the projection direction was chosen using the class labels (Healthy and Diseased). The resulting LDA projection maximized the separation between the two classes. After projection, the Healthy samples and Diseased samples were completely separated by a single threshold, demonstrating perfect linear separability.

## Conclusion

PCA is primarily used to reduce the dimensionality of data while preserving as much variance as possible, regardless of class information. In contrast, LDA uses class labels to find a projection that best separates different classes, making it particularly suitable for classification tasks.

````python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# -----------------------------
# Create the dataset
# -----------------------------
healthy = np.array([
    [2, 3],
    [3, 4],
    [4, 3],
    [3, 2]
])

diseased = np.array([
    [7, 8],
    [8, 9],
    [9, 8],
    [8, 7]
])

# Combine data
X = np.vstack((healthy, diseased))

# Class labels
y = np.array([0]*4 + [1]*4)

# -----------------------------
# Display results
# -----------------------------

np.set_printoptions(precision=6, suppress=True)

print("="*60)
print("Original Dataset")
print("="*60)
print(X)
print("_"*60)
print("\nClass Labels")
print(y)
````

````python
# -----------------------------
# Fit the LDA model
# -----------------------------
lda = LinearDiscriminantAnalysis(n_components=1)
X_lda = lda.fit_transform(X, y)

# -----------------------------
# Display results
# -----------------------------
np.set_printoptions(precision=6, suppress=True)
print("\n" + "="*60)
print("LDA Projection Vector (Scikit-learn)")
print("="*60)
print(lda.scalings_)

print("\n" + "="*60)
print("Projected Samples")
print("="*60)

for i, value in enumerate(X_lda.flatten()):
    cls = "Healthy" if y[i] == 0 else "Diseased"
    print(f"Sample {i+1} ({cls}): {value:.6f}")

````

````python
# -----------------------------
# Plot projected points
# -----------------------------
plt.figure(figsize=(8,2.5))

plt.scatter(
    X_lda[y==0],
    np.zeros(np.sum(y==0)),
    s=100,
    label="Healthy"
)

plt.scatter(
    X_lda[y==1],
    np.ones(np.sum(y==1)),
    s=100,
    label="Diseased"
)

plt.yticks([0,1], ["Healthy","Diseased"])
plt.xlabel("Projection onto LDA Axis")
plt.title("LDA Projection of Samples")
plt.grid(False)
plt.legend()

plt.show()
````

## <font face="Georgia" size="5" color="black"> Comparision: Manual vs. Scikit Implementation

<font face="Georgia" size="3" color="black">

Although the projected values obtained manually and those computed by scikit-learn are numerically different, they represent the same LDA solution. The differences are due to data centering and scaling of the projection vector used by scikit-learn. These transformations do not affect the projection direction, the relative positions of the samples, or the classification results.
The above can be further verified by comparing the plots:

````python
import numpy as np
import matplotlib.pyplot as plt

# Manual projection values
healthy_manual = np.array([3.535, 4.949, 4.949, 3.535])
diseased_manual = np.array([10.605, 12.019, 12.019, 10.605])

# Scikit-learn projection values
healthy_sklearn = np.array([-5.196152, -3.464102, -3.464102, -5.196152])
diseased_sklearn = np.array([3.464102, 5.196152, 5.196152, 3.464102])

fig, axes = plt.subplots(1, 2, figsize=(12, 3))

# -----------------------------
# Manual Projection
# -----------------------------
axes[0].scatter(
    healthy_manual,
    np.zeros_like(healthy_manual),
    s=100,
    label="Healthy"
)

axes[0].scatter(
    diseased_manual,
    np.ones_like(diseased_manual),
    s=100,
    label="Diseased"
)

axes[0].axvline(
    7.777,
    linestyle="--",
    linewidth=2,
    label="Decision Threshold"
)

axes[0].set_yticks([0, 1])
axes[0].set_yticklabels(["Healthy", "Diseased"])
axes[0].set_xlabel("Projection")
axes[0].set_title("Manual LDA Projection")
axes[0].grid(False)
axes[0].legend()

# -----------------------------
# Scikit-learn Projection
# -----------------------------
axes[1].scatter(
    healthy_sklearn,
    np.zeros_like(healthy_sklearn),
    s=100,
    label="Healthy"
)

axes[1].scatter(
    diseased_sklearn,
    np.ones_like(diseased_sklearn),
    s=100,
    label="Diseased"
)

# Threshold midway between the two classes
threshold = (-3.464102 + 3.464102) / 2

axes[1].axvline(
    threshold,
    linestyle="--",
    linewidth=2,
    label="Decision Threshold"
)

axes[1].set_yticks([0, 1])
axes[1].set_yticklabels(["Healthy", "Diseased"])
axes[1].set_xlabel("Projection")
axes[1].set_title("Scikit-learn LDA Projection")
axes[1].grid(False)
axes[1].legend()

plt.tight_layout()
plt.show()
````

## <font face="Georgia" size="5" color="black"> Task 4 - Compute Explained Variance Ratio:

<font face="Georgia" size="3" color="black">

````python
import numpy as np
import matplotlib.pyplot as plt

# Dataset
healthy = np.array([
    [2,3],
    [3,4],
    [4,3],
    [3,2]
])

diseased = np.array([
    [7,8],
    [8,9],
    [9,8],
    [8,7]
])

# Normalized LDA projection vector
w = np.array([0.707, 0.707])

# Threshold (midpoint of projected class means)
threshold = 7.777

plt.figure(figsize=(6,6))

plt.scatter(healthy[:,0], healthy[:,1], label='Healthy', s=80)
plt.scatter(diseased[:,0], diseased[:,1], label='Diseased', s=80)

# Decision boundary:
# w^T x = threshold
# 0.707x + 0.707y = threshold

x = np.linspace(1,10,100)
y = (threshold - w[0]*x)/w[1]

plt.plot(x, y, '--', linewidth=2, label='LDA Decision Boundary')

# Projection vector
plt.quiver(5,5,
           w[0], w[1],
           angles='xy',
           scale_units='xy',
           scale=0.2,
           width=0.008)

plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("LDA Decision Boundary")
plt.legend()
plt.grid(True)

plt.show()
````
