## <font face="Georgia" size="5" color="black"> Task 4 - Effect of threshold on FP and FN.

<font face="Georgia" size="3" color="black">

The classification threshold determines how the predicted probability is converted into a class label.

- A **lower threshold** (e.g., **0.5**) predicts more samples as **Default (Class 1)**.
- A **higher threshold** (e.g., **0.7**) predicts fewer samples as **Default (Class 1)**.

The effect on classification errors is summarized below:

| Threshold | False Positives (FP) | False Negatives (FN) |
|-----------:|:--------------------:|:--------------------:|
| Lower (0.5) | Increases | Decreases |
| Higher (0.7) | Decreases | Increases |

## Conclusion

Increasing the threshold makes the model more conservative when predicting the **Default** class. This reduces **False Positives** but may increase **False Negatives**. Conversely, lowering the threshold reduces **False Negatives** but may result in more **False Positives**.

## <font face="Georgia" size="5" color="black"> Task 5 - Why cross - entropy loss is used instead of MSE.

<font face="Georgia" size="3" color="black">

Cross-Entropy Loss is preferred over Mean Squared Error (MSE) for Logistic Regression because it is specifically designed for binary classification problems.

Some key reasons are:

- It directly measures the difference between the predicted probabilities and the true class labels.
- It penalizes incorrect predictions more heavily, especially when the model is highly confident but wrong.
- It produces a convex optimization problem for logistic regression, making it easier to find the global minimum.
- It generally leads to faster convergence and better classification performance than MSE.

## Conclusion

Therefore, **Cross-Entropy Loss** is preferred over **MSE** because it is better suited for probabilistic classification, provides stronger learning signals, and results in more effective optimization of the logistic regression model.

## <font face="Georgia" size="5" color="black"> Task 6 - State two assumptions of Logistic Regerssion

<font face="Georgia" size="3" color="black">

Some important assumptions of Logistic Regression are:

1. **Independent Observations:**  
   Each observation is assumed to be independent of the others.

2. **Linear Relationship with the Log-Odds:**  
   The independent variables are assumed to have a linear relationship with the **log-odds (logit)** of the dependent variable.

## Conclusion

If these assumptions are reasonably satisfied, Logistic Regression can effectively model the probability of a binary outcome.
