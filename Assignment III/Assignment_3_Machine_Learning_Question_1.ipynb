### Mathematical Solution

#### 1. Standardize the Dataset

We standardize each feature using the formula:

$$\mathbf{ Z = \frac{X - \mu}{\sigma}} $$

Where:
- $\mathbf{\mu}$ = mean of the feature
- $\mathbf{\sigma}$ = standard deviation of the feature (using **population standard deviation**, i.e., dividing by $n$, since we treat this as the full dataset)

---

## Step 1: Compute Mean ($\mu$) for Each Feature

**Experience (Years):**
$$\mathbf{ \mu_{exp} = \frac{2 + 4 + 5 + 6 + 8}{5} = \frac{25}{5} = 5} $$

**Technical Score:**
$$\mathbf{ \mu_{tech} = \frac{65 + 80 + 85 + 90 + 98}{5} = \frac{418}{5} = 83.6} $$

**Communication Score:**
$$\mathbf{ \mu_{comm} = \frac{70 + 78 + 82 + 88 + 95}{5} = \frac{413}{5} = 82.6} $$

<center>Thus,mean are: $\mathbf{\mu = [5, 83.6, 82.6]}$</center>

---

## Step 2: Compute Variances and Standard Deviations

**Experience:**
$$ \sigma_{exp}^2 = \frac{(2-5)^2 + (4-5)^2 + (5-5)^2 + (6-5)^2 + (8-5)^2}{5} = \frac{9+1+0+1+9}{5} = \frac{20}{5} = 4 $$

$$ \sigma_{exp} = \sqrt{4} = 2 $$

**Technical Score:**
$$ \sigma_{tech}^2 = \frac{(65-83.6)^2 + (80-83.6)^2 + (85-83.6)^2 + (90-83.6)^2 + (98-83.6)^2}{5} $$

$$ \sigma_{tech}^2 = \frac{345.96 + 12.96 + 1.96 + 40.96 + 207.36}{5} = \frac{609.2}{5} = 121.84 $$

$$ \sigma_{tech} = \sqrt{121.84} \approx 11.0381 $$

**Communication Score:**
$$ \sigma_{comm}^2 = \frac{(70-82.6)^2 + (78-82.6)^2 + (82-82.6)^2 + (88-82.6)^2 + (95-82.6)^2}{5} $$

$$ \sigma_{comm}^2 = \frac{158.76 + 21.16 + 0.36 + 29.16 + 153.76}{5} = \frac{363.2}{5} = 72.64 $$

$$ \sigma_{comm} = \sqrt{72.64} \approx 8.5229 $$

---

## <font face="Georgia" size="5" color="black"> Task 2 - Compute Covariance Matrix:
For standardized data, the covariance matrix is computed as

$$\mathbf{
C=\frac{1}{n-1}X^TX}
$$

where

- $\mathbf{n=5}$ observations,
- $\mathbf{X}$ is the standardized data matrix.

Since there are **5 observations**,

$$\mathbf{
n-1=5-1=4.
}$$

Therefore,

$$\mathbf{
C=\frac{1}{4}X^TX.
}
$$

---

$$\mathbf{
X=
\begin{bmatrix}

0.1268 & -0.0704 & 0.0613\\
0.5798 & 0.6336 & 0.8269\\
1.3046 & 1.4549 & 1.2862
\end{bmatrix}
}$$

The transpose of $\mathbf{X}$ is

$$\mathbf{
X^T=
\begin{bmatrix}
-1.6851 & -0.3261 & 0.1268 & 0.5798 & 1.3046\\

-1.4699 & -0.7044 & 0.0613 & 0.8269 & 1.2862
\end{bmatrix}
}$$

---
## Step 2. Compute Each Element of $\mathbf{X^TX}$

Each element is obtained using the dot product of two columns.
Thus,
$$\mathbf{

3.9190 & 3.9864 & 3.9998

Since

$$\mathbf{

divide every element by 4.

## Final Covariance Matrix
$$\mathbf{

## Interpretation
- **Symmetric:** $\mathbf{C_{12} = C_{21} = 0.9874}$ ✓

- Since the variables are highly correlated, much of the variation in the dataset can be captured by a small number of principal components. This makes the dataset well suited for **Principal Component Analysis (PCA)**. ✓

## Step 1. Find the Eigenvalues

The eigenvalues are obtained by solving the characteristic equation

$$\mathbf{
\det(C-\lambda I)=0
}$$

where

- $\mathbf{\lambda}$ is an eigenvalue,
- $\mathbf{I}$ is the $\mathbf{3\times3}$ identity matrix.

Using the approximated covariance matrix,

$$\mathbf{
\left|
\begin{array}{ccc}
1.0000-\lambda & 0.9874 & 0.9798\\
0.9874 & 1.0000-\lambda & 0.9966\\
0.9798 & 0.9966 & 1.0000-\lambda
\end{array}
\right|=0.
}$$

Expanding the determinant along the first row,

$$\mathbf{
\begin{aligned}
&=
(1.0000-\lambda)
\begin{vmatrix}
1.0000-\lambda & 0.9966\\
0.9966 & 1.0000-\lambda
\end{vmatrix} \\
&\quad
-0.9874
\begin{vmatrix}
0.9874 & 0.9966\\
0.9798 & 1.0000-\lambda
\end{vmatrix} \\
&\quad
+0.9798
\begin{vmatrix}
0.9874 & 1.0000-\lambda\\
0.9798 & 0.9966
\end{vmatrix}
=0.
\end{aligned}
}$$

Evaluating each $2\times2$ determinant,

$$\mathbf{
\begin{aligned}
&(1.0000-\lambda)\left[(1.0000-\lambda)^2-(0.9966)^2\right]\\
&\quad-0.9874\left[0.9874(1.0000-\lambda)-0.9966(0.9798)\right]\\
&\quad+0.9798\left[0.9874(0.9966)-(1.0000-\lambda)(0.9798)\right]
=0.
\end{aligned}
}$$


---

Expanding and simplifying the above equation,

$$\mathbf{
\begin{aligned}
&(1.0000-\lambda)\left(\lambda^2-2.0000\lambda+0.0068\right)\\
&\quad-0.9874\left(0.0080-0.9874\lambda\right)\\
&\quad+0.9798\left(0.0067+0.9798\lambda\right)=0.
\end{aligned}
}$$

Expanding all the terms,

$$\mathbf{
\begin{aligned}
&-\lambda^3
+3.0000\lambda^2
-2.0068\lambda
+0.0068\\
&\quad-0.0079
+0.9750\lambda\\
&\quad+0.0066
+0.9600\lambda
=0.
\end{aligned}
}$$

Combining like terms,

$$\mathbf{
-\lambda^3
+3.0000\lambda^2
-0.0718\lambda
+0.0055
=0.
}$$

Multiplying both sides by \(-1\),

$$\mathbf{
\boxed{
\lambda^3
-3.0000\lambda^2
+0.0718\lambda
-0.0055
=0.
}
}$$

The roots of the characteristic polynomial are the eigenvalues of the covariance matrix.

The eigenvalues are

$$\mathbf{
\boxed{

## Step 1: Characteristic Equation

The eigenvalues are obtained by solving the characteristic equation

$$\mathbf{
\det(C-\lambda I)=0
}$$

where

- $\mathbf{\lambda}$ is an eigenvalue,
- $\mathbf{I}$ is the $\mathbf{3\times3}$ identity matrix.

Using the approximated covariance matrix,
To find eigenvalues, we solve:

$$\mathbf{ |\mathbf{C} - \lambda\mathbf{I}| = 0 }$$

$$\mathbf{ \begin{vmatrix}
1.000 - \lambda & 0.987 & 0.980 \\
0.987 & 1.000 - \lambda & 0.997 \\
0.980 & 0.997 & 1.000 - \lambda
\end{vmatrix} = 0 }$$

---

## Step 2: Expand the Determinant

Using the formula for a 3×3 determinant:

$$ (1-\lambda)[(1-\lambda)(1-\lambda) - (0.997)(0.997)] - 0.987[0.987(1-\lambda) - (0.997)(0.980)] + 0.980[0.987(0.997) - (1-\lambda)(0.980)] = 0 $$

---

## Step 3: Simplify

$$ (1-\lambda)[(1-\lambda)^2 - 0.994009] - 0.987[0.987(1-\lambda) - 0.97706] + 0.980[0.984039 - 0.980(1-\lambda)] = 0 $$

$$ (1-\lambda)[(1-\lambda)^2 - 0.994009] - 0.987[0.987 - 0.987\lambda - 0.97706] + 0.980[0.984039 - 0.980 + 0.980\lambda] = 0 $$

$$ (1-\lambda)[(1-\lambda)^2 - 0.994009] - 0.987[0.00994 - 0.987\lambda] + 0.980[0.004039 + 0.980\lambda] = 0 $$

$$ (1-\lambda)[(1-\lambda)^2 - 0.994009] - 0.009811 + 0.974169\lambda + 0.003958 + 0.9604\lambda = 0 $$

$$ (1-\lambda)[(1-\lambda)^2 - 0.994009] + 1.934569\lambda - 0.005853 = 0 $$

---

## Step 4: Expand Further

$$ (1-\lambda)[(1 - 2\lambda + \lambda^2) - 0.994009] + 1.934569\lambda - 0.005853 = 0 $$

$$ (1-\lambda)[0.005991 - 2\lambda + \lambda^2] + 1.934569\lambda - 0.005853 = 0 $$

$$ (1-\lambda)(\lambda^2 - 2\lambda + 0.005991) + 1.934569\lambda - 0.005853 = 0 $$

$$ (\lambda^2 - 2\lambda + 0.005991) - \lambda(\lambda^2 - 2\lambda + 0.005991) + 1.934569\lambda - 0.005853 = 0 $$

$$ \lambda^2 - 2\lambda + 0.005991 - \lambda^3 + 2\lambda^2 - 0.005991\lambda + 1.934569\lambda - 0.005853 = 0 $$

$$ -\lambda^3 + 3\lambda^2 + (-2 + 1.934569 - 0.005991)\lambda + (0.005991 - 0.005853) = 0 $$

$$ -\lambda^3 + 3\lambda^2 - 0.071422\lambda + 0.000138 = 0 $$

---

## Step 5: Solve for λ

$$ \lambda^3 - 3\lambda^2 + 0.071422\lambda - 0.000138 = 0 $$

Solving this cubic equation the eigen values which are:

$$\mathbf{

## <font face="Georgia" size="5" color="black"> Task 7 - State one advantage and one limitation of PCA

<font face="Georgia" size="3" color="black">

## Advantage

One major advantage of PCA is that it **reduces the dimensionality of a dataset while retaining most of the important information**. This simplifies data analysis, reduces computational complexity, and can improve the efficiency of many machine learning algorithms.

---

## Limitation

One limitation of PCA is that **the transformed principal components are linear combinations of the original variables**, making them more difficult to interpret. Additionally, PCA captures only linear relationships and may not effectively represent nonlinear patterns in the data.

---

## <font face="Georgia" size="5" color="black"> Implement PCA using scikit-learn.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# --------------------------------------------
# Step 1: Original Dataset
# --------------------------------------------

employees = ["A", "B", "C", "D", "E"]

X = np.array([
    [2, 65, 70],
    [4, 80, 78],
    [5, 85, 82],
    [6, 90, 88],
    [8, 98, 95]
])

print("Original Dataset")
print(X)
```

```python
# --------------------------------------------
# Step 2: Standardize the Data
# --------------------------------------------

scaler = StandardScaler()

X_std = scaler.fit_transform(X)

print("\nStandardized Dataset")
print(np.round(X_std, 4))
```

<font face="Georgia" size="3" color="black">

The standardized values produced by $\texttt{StandardScaler}$ closely match the standardized data matrix computed manually.

$$
\mathbf{Z}=
\begin{bmatrix}
-1.5000 & -1.6851 & -1.4784\\
-0.5000 & -0.3261 & -0.5397\\
0.0000 & 0.1268 & -0.0704\\
0.5000 & 0.5798 & 0.6336\\
1.5000 & 1.3046 & 1.4550
\end{bmatrix}.
$$

Minor differences may occur because Python retains full floating-point precision, whereas the manual calculations were rounded to four decimal places.
</font>

---

```python
# --------------------------------------------
# Step 3: Perform PCA
# --------------------------------------------

pca = PCA(n_components=3)

X_pca = pca.fit_transform(X_std)

# --------------------------------------------
# Step 4: Display PCA Results
# --------------------------------------------

print("\nEigenvalues")
print(np.round(pca.explained_variance_, 6))

print("\nExplained Variance Ratio")
print(np.round(pca.explained_variance_ratio_, 6))

print("\nPrincipal Components (Eigenvectors)")
print(np.round(pca.components_, 6))

print("\nReduced Dataset")
print(np.round(X_pca, 4))
```

<font face="Georgia" size="3" color="black">

Comparison: The eigenvalues obtained using scikit-learn differ from the manually computed values because the manual solution used a normalized covariance (correlation) matrix with unit variances, whereas scikit-learn computes PCA using the sample covariance matrix of the standardized data. Consequently, the Python eigenvalues are scaled by a constant factor. Despite this difference, both approaches identify the same dominant principal component, and the explained variance ratios and overall conclusions remain consistent.
</font>

---

```python
# --------------------------------------------
# Step 5: Plot Original Standardized Data
# --------------------------------------------

fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    X_std[:,0],
    X_std[:,1],
    X_std[:,2],
    s=80
)

for i, name in enumerate(employees):
    ax.text(
        X_std[i,0],
        X_std[i,1],
        X_std[i,2],
        name,
        fontsize=10
    )

ax.set_title("Original Standardized Employee Data")
ax.set_xlabel("Experience")
ax.set_ylabel("Technical Score")
ax.set_zlabel("Communication Score")

plt.show()
```

```python
# --------------------------------------------
# Step 6: Plot PCA Reduced Data
# --------------------------------------------

plt.figure(figsize=(7,6))

plt.scatter(
    X_pca[:,0],
    X_pca[:,1],
    s=80
)

for i, name in enumerate(employees):
    plt.text(
        X_pca[i,0],
        X_pca[i,1],
        name,
        fontsize=10
    )

plt.title("Employee Data After PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)

plt.show()
```
